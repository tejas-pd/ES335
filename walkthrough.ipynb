```python
import pandas as pd
m = json.load(open('results/metrics.json'))
rows = {'random split': {k: v['mae']/1e5 for k, v in m['random_split']['overall'].items()}}
for s in 'ABCD':
    rows[f'held-out {s}'] = {k: v['mae']/1e5 for k, v in m['loso'][s]['overall'].items()}
pd.DataFrame(rows).T.round(2)  # MAE in lakh
```

```python
pd.DataFrame({s: {k: v['bias_pct'] for k, v in m['loso'][s]['overall'].items()} for s in 'ABCD'}).round(1)  # median signed % error on held-out sources
```
